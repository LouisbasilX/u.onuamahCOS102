```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns



bnb = pd.read_csv("NYC_airbnb.csv")
emp = ["host_name", "name"]
for e in emp:
    bnb[e] = bnb[e].fillna(bnb[e].mode()[0])

```

```python


bnb.dropna(inplace = True)
bnb.isna().sum()
```

```python
#A Heatmap of the data

plt.figure(figsize = (10,10))
corr = bnb.select_dtypes(include=['number']).corr()
sns.heatmap(corr, annot = True, cmap = "coolwarm")
plt.title("Bnb Heat-map")
plt.show()
```

**Heatmap Observations:** 
<ul>
<li>ID and Host_id are highly correlated</li>
<li>Number of reviews and Reviews per month are highly related too</li>
<li>Most other data show little to no similarities</li>
</ul>


```python
#A Count plot of the data
plt.figure(figsize=(10, 6))
# Sort them so the highest borough is first
sns.countplot(x='neighbourhood_group', data=bnb, order=bnb['neighbourhood_group'].value_counts().index)
plt.title("Number of Listings per Borough (Neighbourhood Group)")
plt.xlabel("Borough")
plt.ylabel("Count")
plt.savefig("listing_counts.png")
```

**Countplot Observations:** 
<ul>
