```python
def changeme(mylist):
    mylist.append([1,2,3,4])
    print("inside mylist", mylist)
    return
mylist = [10,3,4]
changeme(mylist)
print("outside mylist", mylist)
```

```python
def changeme(mylist):
    mylist = [1,2,3,4]
    print("Values inside the function: ", mylist)
    return
mylist = [10,3,5]
changeme(mylist)
print("Values outside the function: ", mylist)
```

```python
def printinfo(name, age):
    print("Name: ", name)
    print("Age: ", age)
    return
printinfo(age = 50, name = "miki")
```

```python
def printinfo(name, age = 35):
    print("Name: ", name)
    print("Age: ", age)
    return
printinfo(age=50, name="miki")
printinfo(name="miki")
```

```python
def printinfo(arg1, *tup):
    print("output is: ")
    print(arg1)
    for var in tup:
        print(var)
        return
printinfo(10)
printinfo(70, 60, 50)
```

```python
total = 50
def sum(arg1, arg2):
    total = arg1 + arg2
    print("Inside the function local total: ", total)
    return total
sum(10,20)
print("outside the function global total: ", total)
```
