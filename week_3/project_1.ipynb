```python
import tkinter as tk
from tkinter import messagebox as mb
import csv

employees  = set()
try:
    with open('GIG-logistics.csv', mode = 'r') as file:
        csv_reader = csv.reader(file)
        next(csv_reader)

        for row in csv_reader:
            employees.add(f"{row[0]}, {row[1]}")
    print("converted successfully")
except FileNotFoundError:
    print("File not found")
print(employees)                

def is_employee(name, dept):
    name = name.lower().strip()
    dept = dept.lower().strip()
    query = f"{name}, {dept}"
    return query in employees
def welcome_message(name, Department):
    window = tk.Toplevel(root)
    window.title("Welcome page")
    window.geometry("500x500")

    label_1 = tk.Label(window, text = f"Welcome {name}")
    label_1.pack()
    label_2 = tk.Label(window, text = f"Department {Department}")
    label_2.pack()

    root.mainloop()
def submit():
    firstname = firstname_entry.get()
    lastname = lastname_entry.get()
    dept = dept_entry.get()
    name = f"{firstname} {lastname}"
    if is_employee(name, dept):
        welcome_message(name, dept)
    else:
       mb.showerror("Validation", "invalid Employee entry")    


root = tk.Tk()
root.title("Login Form")
root.geometry("500x200")

firstname_label = tk.Label(root, text="firstname")
firstname_label.pack()
firstname_entry = tk.Entry(root)
firstname_entry.pack()

lastname_label = tk.Label(root, text="lastname")
lastname_label.pack()
lastname_entry = tk.Entry(root)
lastname_entry.pack()

dept_label  = tk.Label(root, text = "Department")
dept_label.pack()
dept_entry = tk.Entry(root)
dept_entry.pack()

submit_button = tk.Button(root, text = "Submit", command = submit)
submit_button.pack()

root.mainloop()



    

```
